### Matplotlib -  Scatter Plots

```python
## Import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
## Set style and inline plots
%matplotlib inline
plt.style.use('seaborn-whitegrid')
```

```python
## plt.plot/ax.plot work with scatter plots as well as line plots
x = np.linspace(0, 10, 30)
y = np.sin(x)

plt.plot(x, y, 'o', color='black'); # Third arguement is the function call for the type of symbol used in plotting
```

```python
## Examples of other symbols that can be used in scatter plots
rng = np.random.RandomState(0)
for marker in ['o', '.', ',', 'x', '+', 'v', '^', '<', '>', 's', 'd']:
    plt.plot(rng.rand(5), rng.rand(5), marker, 
    label="marker='{0}'".format(marker))

    plt.legend(numpoints=1)
    plt.xlim(0, 1.8);
```

```python
## Can be used in conjuction with line and color codes
plt.plot(x, y, '-ok'); # line (-), circle marker (o), black (k)
```

```python
plt.plot(x, y, '-p', color='gray', markersize=15, linewidth=4, markerfacecolor='white',
markeredgecolor='gray', markeredgewidth=2);
```

```python
## Second, more powerful method of creating scatter plot is plt.scatter function
plt.scatter(x, y, marker='o');
```

Primary difference of plt.scatter from plt.plot is that it can be used to create scatter plots 
where is the properties of each individual point can be controlled or mapped to data

```python
## Size argument is given in pixels

rng = np.random.RandomState(0)
x = rng.randn(100)
y = rng.randn(100)
colors = rng.rand(100)
sizes = 1000 * rng.rand(100)

plt.scatter(x, y, c=colors, s=sizes, alpha=0.3, cmap='viridis') # alpha controls transparency
plt.colorbar(); # show color scale
```

### plot Versus scatter: A Note on Efficiency

For datasets larger than a few thousand points, plt.plot can be noticeably more efficient than plt.scatter.

plt.scatter must render each point individually.

plt.plot should be preferred over plt.scatter for large datasets.
