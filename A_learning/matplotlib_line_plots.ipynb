#### MATLAB-style interface

MATLAB-style tools are contained in the pyplot (plt) interface.

It’s important to note that this interface is stateful: it keeps track of the “current” figure
and axes, which are where all plt commands are applied. You can get a reference to these using 
the plt.gcf() (get current figure) and plt.gca() (get current axes) routines.

While this stateful interface is fast and convenient for simple plots, it is easy to run
into problems. Changing something on first panel after second is created is possible, but clunky.
See Object-Oriented interface for better solution.

```python
x = np.linspace(0, 10, 100)

plt.figure() # create a plot figure

# create the first of two panels and set current axis
plt.subplot(2, 1, 1) # (rows, columns, panel number)
plt.plot(x, np.sin(x))

# create the second panel and set current axis
plt.subplot(2, 1, 2)
plt.plot(x, np.cos(x));
```

#### Object-oriented interface

Object-oriented interface is available for more complicated situations and more control over
figure is wanted.

Rather than depending on some notion of an “active” figure or axes, in the object-oriented interface 
the plotting functions are methods of explicit Figure and Axes objects.



```python
## Recreate previous plot using object-oriented interface
## First create a grid of plots
## ax will be an array of the two Axes objects
fig, ax = plt.subplots(2)

## Call plot() method on the appropriate object
ax[0].plot(x, np.sin(x))
ax[1].plot(x, np.cos(x));
```

#### Simple line plot

```python
## Set style
plt.style.use('seaborn-whitegrid')
```

```python
## For all Matplotlib plots, start by creating a figure and an axes
fig = plt.figure() #single container that contains all the objects representing axes, graphics, text, and labels
ax = plt.axes() #bounding box with ticks and labels, which will eventually contain the plot elements that make the visualization
```

```python
## Plot a simple sinusoid
fig = plt.figure()
ax = plt.axes()

x = np.linspace(0, 10, 1000)
ax.plot(x, np.sin(x));
```

```python
## Plot a simple sinusoid with pylab interface
plt.plot(x, np.sin(x));
```

```python
## For figure with multiple line, call plot function several times
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x));
```

```python
## Adjusting color of lines
plt.plot(x, np.sin(x - 0), color='blue') # specify color by name
plt.plot(x, np.sin(x - 1), color='g') # short color code (rgbcmyk)
plt.plot(x, np.sin(x - 2), color='0.75') # Grayscale between 0 and 1
plt.plot(x, np.sin(x - 3), color='#FFDD44') # Hex code (RRGGBB from 00 to FF)
plt.plot(x, np.sin(x - 4), color=(1.0,0.2,0.3)) # RGB tuple, values 0 and 1
plt.plot(x, np.sin(x - 5), color='chartreuse'); # all HTML color names supported
```

```python
## Adjust line style
plt.plot(x, x + 0, linestyle='solid')
plt.plot(x, x + 1, linestyle='dashed')
plt.plot(x, x + 2, linestyle='dashdot')
plt.plot(x, x + 3, linestyle='dotted');

## For short, you can use the following codes:
plt.plot(x, x + 4, linestyle='-') # solid
plt.plot(x, x + 5, linestyle='--') # dashed
plt.plot(x, x + 6, linestyle='-.') # dashdot
plt.plot(x, x + 7, linestyle=':'); # dotted
```

```python
## For extremely terse code, linestyle and color codes can be combined into a single nonkeyword argument
plt.plot(x, x + 0, '-g') # solid green
plt.plot(x, x + 1, '--c') # dashed cyan
plt.plot(x, x + 2, '-.k') # dashdot black
plt.plot(x, x + 3, ':r'); # dotted red
```

### Adjusting the Plot
#### Axes Limits

```python
## Set axes limits with xlim and ylim
plt.plot(x, np.sin(x))

plt.xlim(-1, 11)
plt.ylim(-1.5, 1.5);
```

```python
## Reversing the axes
plt.plot(x, np.sin(x))

plt.xlim(10, 0)
plt.ylim(1.2, -1.2);
```

```python
## plt.axis() method allows setting of x and y limits in a single call
## plt.axis([xmin, xmax, ymin, ymax])
plt.plot(x, np.sin(x))
plt.axis([-1, 11, -1.5, 1.5]);
```

```python
## plt.axis() can automatically tighten the bounds around the current plot
plt.plot(x, np.sin(x))
plt.axis('tight');
```

```python
## Set equal aspect ratio, so on screen one unit of x is equal to one unit of y
plt.plot(x, np.sin(x))
plt.axis('equal');
```

```python
## Labeling Plots
## Can futher customize, look at Matplotlib documentation
plt.plot(x, np.sin(x))
plt.title("A Sine Curve")
plt.xlabel("x")
plt.ylabel("sin(x)");
```

```python
## Legends
plt.plot(x, np.sin(x), '-g', label='sin(x)')
plt.plot(x, np.cos(x), ':b', label='cos(x)')
plt.axis('equal')

plt.legend();
```

#### Matplotlib Gotchas

Most plt functions translate directly to ax methods, but this is not always the case.

+ plt.xlabel() → ax.set_xlabel()
+ plt.ylabel() → ax.set_ylabel()
+ plt.xlim() → ax.set_xlim()
+ plt.ylim() → ax.set_ylim()
+ plt.title() → ax.set_title()

```python
## For Object-oriented interface, set all properties at once rather than calling functions individually
ax = plt.axes()
ax.plot(x, np.sin(x))
ax.set(xlim=(0, 10), ylim=(-2, 2),
xlabel='x', ylabel='sin(x)',
title='A Simple Plot');
```
